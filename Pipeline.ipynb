Using the OAuth we are sending a login request by opening the browser and once if any user gives access then we are with that user.


```python
import os
from dotenv import load_dotenv
import spotipy
from spotipy.oauth2 import SpotifyOAuth

print("Loaded .env:", load_dotenv())

sp = spotipy.Spotify(auth_manager=SpotifyOAuth(
    scope="playlist-read-private playlist-read-collaborative",
    open_browser=True,
    cache_path=".spotify_token_cache",
))

me = sp.current_user()
print("Logged in as:", me["display_name"], "|", me["id"])
```

```python
# current_user_playlists will give you the playlists of the user who logged in.
results = sp.current_user_playlists(limit=50)
for p in results["items"]:
    print(p["id"], "|", p["name"], "| owner:", p["owner"]["id"])
```

# Extracting Album

```python
album = []

for item in results['items']:
    album_id = item['item']['album']['id']
    album_name = item['item']['album']['name']
    album_release_date = item['item']['album']['release_date']
    album_total_tracks = item['item']['album']['total_tracks']
    album_spotify_url = item['item']['external_urls']['spotify']

    album_elements = {
        'album_id': album_id,
        'album_name': album_name,
        'album_release_date': album_release_date,
        'album_total_tracks': album_total_tracks,
        'album_spotify_url': album_spotify_url
    }
    album.append(album_elements)
```

```python
print(album)
```

# Extracting Artists

```python
artists = []

for row in results['items']:
    for key,value in row.items():
        if key == 'item':
            for artist in value['artists']:
                artist_dict = {
                    'artist_id': artist['id'],
                    'artist_name': artist['name'],
                    'external_url': artist['href']
                }
                artists.append(artist_dict)
print(artists)
```

# Extracting Songs

```python
song_list = []

for row in results['items']:
    song_id = row['item']['id']
    song_name = row['item']['name']
    song_duration = row['item']['duration_ms']
    song_url = row['item']['external_urls']['spotify']
    song_added_at = row['added_at']
    album_id = row['item']['album']['id']
    artist_id = row['item']['artists'][0]['id']

    song_dict = {
        'song_id': song_id,
        'song_name': song_name,
        'song_duration': song_duration,
        'song_url': song_url,
        'song_added_at': song_added_at,
        'album_id': album_id,
        'artist_id': artist_id
    }
    song_list.append(song_dict)

print(song_list)
```

```python
import pandas as pd
album_df = pd.DataFrame(album)
artists_df = pd.DataFrame(artists)
songs_df = pd.DataFrame(song_list)
```

# Transformation

### Remove duplicates

```python
album_df = album_df.drop_duplicates(subset = ['album_id'])
artists_df = artists_df.drop_duplicates(subset = ['artist_id'])
```

### convert string date to Date Time

```python
album_df['album_release_date'] = pd.to_datetime(album_df['album_release_date'], errors='coerce')
```

```python
songs_df['song_added_at'] = pd.to_datetime(songs_df['song_added_at'], errors='coerce')
```
